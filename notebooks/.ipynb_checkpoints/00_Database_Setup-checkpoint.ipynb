# Database Creation and Data Loading

## Objective

This notebook creates the Mutual Fund Analytics SQLite database using the schema.sql file and loads all cleaned datasets into their respective tables.

### Tables Created

Dimension Tables
- dim_fund
- dim_date

Fact Tables
- fact_nav
- fact_transactions
- fact_performance
- fact_aum

### Data Sources

- clean_01_fund_master.csv
- clean_02_nav_history.csv
- clean_03_aum_by_fund_house.csv
- clean_07_scheme_performance.csv
- clean_08_investor_transactions.csv

```python
import sqlite3
import pandas as pd

from pathlib import Path
```

## Define Paths

```python
DATABASE_FILE = "../data/db/Bluestocks_mf.db"

SCHEMA_FILE = "../sql/schema.sql"

DATA_DIR = Path("../data/processed")
```

## Create Database Connection

```python
conn = sqlite3.connect(
    DATABASE_FILE
)

cursor = conn.cursor()

print("Database Connected")
```

## Execute Schema.sql

```python
with open(
    SCHEMA_FILE,
    "r",
    encoding="utf-8"
) as f:

    schema = f.read()

cursor.executescript(schema)

conn.commit()

print("Schema Executed Successfully")
```

## Load Fund Master

```python
dim_fund = pd.read_csv(
    DATA_DIR / "clean_01_fund_master.csv"
)

print(dim_fund.shape)

dim_fund.head()
```

```python
dim_fund.to_sql(
    "dim_fund",
    conn,
    if_exists="append",
    index=False
)

print("dim_fund loaded")
```

## Create Date Dimension

```python
nav = pd.read_csv(
    DATA_DIR / "clean_02_nav_history.csv"
)

nav["date"] = pd.to_datetime(
    nav["date"]
)

date_dim = pd.DataFrame()

date_dim["full_date"] = (
    nav["date"]
    .drop_duplicates()
)

date_dim["date_key"] = (
    date_dim["full_date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)

date_dim["year"] = (
    date_dim["full_date"]
    .dt.year
)

date_dim["quarter"] = (
    date_dim["full_date"]
    .dt.quarter
)

date_dim["month"] = (
    date_dim["full_date"]
    .dt.month
)

date_dim["month_name"] = (
    date_dim["full_date"]
    .dt.month_name()
)

date_dim["day"] = (
    date_dim["full_date"]
    .dt.day
)

date_dim.head()
```

```python
date_dim.to_sql(
    "dim_date",
    conn,
    if_exists="append",
    index=False
)

print("dim_date loaded")
```

## Load NAV Fact Table

```python
fact_nav = pd.read_csv(
    DATA_DIR / "clean_02_nav_history.csv"
)

fact_nav["date"] = pd.to_datetime(
    fact_nav["date"]
)

fact_nav["date_key"] = (
    fact_nav["date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)

fact_nav = fact_nav[
    [
        "amfi_code",
        "date_key",
        "nav"
    ]
]

fact_nav.head()
```

```python
fact_nav.to_sql(
    "fact_nav",
    conn,
    if_exists="append",
    index=False
)

print("fact_nav loaded")
```

## Load Transactions

```python
txn = pd.read_csv(
    DATA_DIR / "clean_08_investor_transactions.csv"
)

txn["transaction_date"] = pd.to_datetime(
    txn["transaction_date"]
)

txn["full_date"] = (
    txn["transaction_date"]
)

txn["date_key"] = (
    txn["transaction_date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)

txn.head()
```

```python
txn.to_sql(
    "fact_transactions",
    conn,
    if_exists="append",
    index=False
)

print("fact_transactions loaded")
```

## Load Performance Data

```python
performance = pd.read_csv(
    DATA_DIR / "clean_07_scheme_performance.csv"
)

performance.head()
```

```python
performance.to_sql(
    "fact_performance",
    conn,
    if_exists="append",
    index=False
)

print("fact_performance loaded")
```

## Load AUM Data

```python
aum = pd.read_csv(
    DATA_DIR / "clean_03_aum_by_fund_house.csv"
)

aum["date"] = pd.to_datetime(
    aum["date"]
)

aum["full_date"] = aum["date"]

aum["date_key"] = (
    aum["date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)

aum.head()
```

```python
aum.to_sql(
    "fact_aum",
    conn,
    if_exists="append",
    index=False
)

print("fact_aum loaded")
```

## Verify Data Loads

```python
tables = [

    "dim_fund",
    "dim_date",
    "fact_nav",
    "fact_transactions",
    "fact_performance",
    "fact_aum"

]

for table in tables:

    count = pd.read_sql(
        f"""
        SELECT COUNT(*)
        AS rows
        FROM {table}
        """,
        conn
    )

    print(
        f"{table}:",
        count.iloc[0,0]
    )
```

```python
# ==================================================
# CREATE DATA DICTIONARY (.md)
# ==================================================

tables = [

    "dim_fund",
    "dim_date",
    "fact_nav",
    "fact_transactions",
    "fact_performance",
    "fact_aum"

]

markdown = []

markdown.append(
    "# Mutual Fund Analytics Platform - Data Dictionary\n"
)

markdown.append(
    "Automatically generated from SQLite database schema.\n"
)

for table in tables:

    markdown.append(f"\n# {table}\n")

    markdown.append(
        "| Column Name | Data Type |\n"
    )

    markdown.append(
        "|------------|-----------|\n"
    )

    schema = pd.read_sql(
        f"""
        PRAGMA table_info({table})
        """,
        conn
    )

    for _, row in schema.iterrows():

        markdown.append(
            f"| {row['name']} | {row['type']} |\n"
        )

# --------------------------------------------------
# SAVE FILE
# --------------------------------------------------

with open(
    "../reports/data_dictionary.md",
    "w",
    encoding="utf-8"
) as f:

    f.writelines(markdown)

print(
    "data_dictionary.md created successfully"
)
```
